### Series
Series is a one-dimensional labeled array capable of holding any data type (integers, strings, 
floating point numbers, 
- Python objects, etc.). The axis labels are collectively referred to as the index.
```
s = pd.Series(data, index=index)
```


````python
import numpy as np
import pandas as pd 

dt = pd.Series(np.random.rand(3), index=["Physics","Chemistry", "Biology"])
dt
````

- output represents normalized scores.

````python
series = pd.Series({
    "Physics": 0.340418,
    "Chemistry": 0.669102,
    "biology": 0.696035
})
percentSeries = series * 100
percentSeries
````

Normalized values 

````python
normalized = series / series.sum()
normalized
````

### Plots a bar chart

````python
import matplotlib.pyplot as plt

series.plot(kind='bar', title='Subject Scores')
plt.ylabel('Score')
plt.show()
````

### ndarray 

- If data is an ndarray, index must be the same length as data. 
- If no index is passed, one will be created having values `[0, ..., len(data) - 1].`

````python
n = pd.Series(np.random.rand(3), index=["Physics","Chemistry", "Biology"])
n
````

### From dict

A series can be instantiated from dicts:

````python
data = {"English": 1, "History": 2, "Geography": 3}
pd.Series(data)
````

````python
pd.Series(data, index=["English", "History", "Geography"])
````
